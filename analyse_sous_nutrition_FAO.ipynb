<h3 style="margin: auto; padding: 20px; color: RGB(0,150,250); ">3.1 - Proportion de personnes en sous nutrition</h3>

```python
# Il faut tout d'abord faire une jointure entre la table population et la table sous nutrition, en ciblant l'année 2017
pop_2017 = population[population['Année'] == 2017]
sous_nutri_2017 = sous_nutrition[sous_nutrition['Année'].str.contains('2016-2018')]
df_2017 = pd.merge(pop_2017, sous_nutri_2017, on='Zone')
df_2017 = df_2017.drop(columns=['Année_y']).rename(columns={'Année_x': 'Année'})
```

```python
#Affichage du dataset
print(df_2017)
```

```python
#Calcul et affichage du nombre de personnes en état de sous nutrition
total_population = df_2017['Population'].sum()
total_sous_nutrition = df_2017['sous_nutrition'].sum()

proportion = (total_sous_nutrition / total_population) * 100

print(f"Nombre total de personnes sous-alimentées en 2017 : {int(total_sous_nutrition):,}")
print(f"Population totale analysée en 2017 : {int(total_population):,}")
print(f"Proportion mondiale de sous-nutrition : {proportion:.2f}%")
```

<div style="border: 1px solid RGB(0,150,250);" >
<h3 style="margin: auto; padding: 20px; color: RGB(0,150,250); ">3.2 - Nombre théorique de personne qui pourrait être nourries</h3>
</div>

```python
#Combien mange en moyenne un être humain ? Source => FAO
besoin_kcal_jour = 2350
```

```python
#On commence par faire une jointure entre le data frame population et Dispo_alimentaire afin d'ajouter dans ce dernier la population
df_pop_di_al = pd.merge(dispo_alimentaire, pop_2017, on='Zone', how='left')
```

```python
#Affichage du nouveau dataframe
print(df_pop_di_al)
```

```python
#Création de la colonne dispo_kcal avec calcul des kcal disponibles mondialement
df_pop_di_al['dispo_kcal'] = (
    df_pop_di_al['Disponibilité alimentaire (Kcal/personne/jour)'] *
    df_pop_di_al['Population'] * 365)
total_kcal_dispo = df_pop_di_al['dispo_kcal'].sum()
print(f"Nombre de kcal disponibles mondialement : {int(total_kcal_dispo):,} kcal")
```

```python
#Calcul du nombre d'humains pouvant être nourris
besoin_kcal_annuel = besoin_kcal_jour * 365
nbr_hum_nourris = total_kcal_dispo / (besoin_kcal_annuel)
print(f"Nombre d'humains pouvant être nourris : {int(nbr_hum_nourris):,} humains")
```

<div style="border: 1px solid RGB(0,150,250);" >
<h3 style="margin: auto; padding: 20px; color: RGB(0,150,250); ">3.3 - Nombre théorique de personne qui pourrait être nourrie avec les produits végétaux</h3>
</div>

```python
#Transfert des données avec les végétaux dans un nouveau dataframe
df_vegetaux = dispo_alimentaire[dispo_alimentaire['Origine'] == 'vegetale']
df_vegetaux.head()
```

```python
#Calcul du nombre de kcal disponible pour les végétaux
dispo_kcal_vegetaux = (df_pop_di_al['Population'] * df_vegetaux['Disponibilité alimentaire (Kcal/personne/jour)'] * 365).sum()
print(f"Le nombre de kcal disponible annuellement pour les végétaux est : {int(dispo_kcal_vegetaux):,} kcal")
```

```python
#Calcul du nombre d'humains pouvant être nourris avec les végétaux
besoin_kcal_annuel= besoin_kcal_jour * 365
nbr_humains_vegetaux = dispo_kcal_vegetaux / besoin_kcal_annuel
print(f"Nombre d'humains pouvant être nourris avec les végétaux annuellement : {int(nbr_humains_vegetaux):,} humains")
```

```python
#Calcul de la disponibilité totale
dispo_interieure_totale = dispo_alimentaire['Disponibilité intérieure'].sum()
print(f"Nombre de disponibilité totale: {int(dispo_interieure_totale):,}")
```

```python
#création d'une boucle for pour afficher les différentes valeurs en fonction des colonnes aliments pour animaux, pertes, nourritures. 
colonnes = ['Aliments pour animaux', 'Pertes', 'Nourriture', 'Semences', 'Traitement', 'Autres Utilisations']
lst_total_col = []

for col in colonnes:
    total = dispo_alimentaire[col].sum()
    lst_total_col.append(total)

somme_total = sum(lst_total_col)
print(somme_total)

for i, tot in enumerate(colonnes):
    proportion = lst_total_col[i]/somme_total * 100
    print(f"Total {tot} : {lst_total_col[i]} soit {proportion:.2f}%")
```

<h3 style="margin: auto; padding: 20px; color: RGB(0,150,250); ">3.5 - Utilisation des céréales</h3>
</div>

```python
#Création d'une liste avec toutes les variables
lst_produit = sorted(dispo_alimentaire['Produit'].unique())
print(lst_produit)
```

```python
#Création d'un dataframe avec les informations uniquement pour ces céréales
cereales = ['Avoine','Blé','Céréales, Autres','Millet','Maïs','Orge','Riz (Eq Blanchi)','Seigle','Sorgho']
df_cereales = dispo_alimentaire[dispo_alimentaire['Produit'].isin(cereales)]
df_cereales.head()
```

```python
#Affichage de la proportion d'alimentation animale
df_animaux = (df_cereales.groupby("Produit")[["Aliments pour animaux", "Disponibilité intérieure"]].sum().reset_index())

tot_aliments_animaux = df_animaux["Aliments pour animaux"]
tot_dispo_alimentaire = df_animaux["Disponibilité intérieure"]

df_animaux["% Aliments pour animaux"] = (tot_aliments_animaux / tot_dispo_alimentaire * 100).round(2)

df_animaux.head()
```

```python
#Affichage de la proportion d'alimentation humaine
df_nourriture = (df_cereales.groupby("Produit")["Nourriture"].sum().reset_index())

tot_nourriture = df_nourriture["Nourriture"]

df_nourriture["% Nourriture"] = (tot_nourriture / tot_dispo_alimentaire * 100).round(2)

df_nourriture.head()
```
